# Suricata Qwen V2.2 — Kaggle Cloud Batch

**Quick start:**
1. Enable GPU: *Settings → Accelerator → GPU T4 x2*
2. Enable Internet access in Kaggle settings
3. Add the exported batch ZIP as a dataset input (or upload it)
4. Click *Run All*

This notebook never connects to the local database. Every successful result is appended and flushed after each rule, so re-running the notebook resumes from the existing `results.jsonl`.

```python
# ============================================================
# CELL 0 — CONFIGURATION
# Rarely needs editing. All paths are auto-discovered.
# ============================================================
OLLAMA_URL = "http://127.0.0.1:11434"
STOP_AFTER_ERRORS = 3
RETRIES = 3
WORK_DIR_PATH = "/kaggle/working/qwen_cloud"
```

```python
# ============================================================
# CELL 1 — GPU CHECK
# Stops immediately if no GPU is detected.
# ============================================================
import subprocess, sys

gpu_check = subprocess.run(["nvidia-smi"], capture_output=True, text=True)
if gpu_check.returncode != 0:
    raise RuntimeError(
        "\n"
        "No NVIDIA GPU is visible.\n"
        "Enable the GPU accelerator in Kaggle Settings before running this batch.\n"
        "Then run this notebook again."
    )
print(gpu_check.stdout[:800])
print("GPU detected.")
```

```python
# ============================================================
# CELL 2 — AUTO-DISCOVER INPUT BATCH
# Searches /kaggle/input recursively for manifest.json.
# Fails safely when 0 or >1 compatible batches are found.
# ============================================================
import shutil, zipfile
from pathlib import Path

input_root = Path("/kaggle/input")
manifests = sorted(input_root.rglob("manifest.json"))

if len(manifests) == 0:
    # Also search for a zip in case the user uploaded to input directly
    zips = sorted(input_root.rglob("cloud_batch_*.zip"))
    if not zips:
        raise FileNotFoundError(
            "\n"
            "No cloud batch was found in /kaggle/input.\n"
            "Add the exported cloud_batch_kaggle-*.zip as a Kaggle dataset input and run again."
        )
    manifests_after_search = []
    # Extract and search inside zips
    for zp in zips:
        with zipfile.ZipFile(zp) as arc:
            if "manifest.json" in arc.namelist():
                manifests_after_search.append(zp)
    if not manifests_after_search:
        raise FileNotFoundError(
            "No compatible cloud batch ZIPs found in /kaggle/input.\n"
            "Add the batch as a Kaggle dataset input."
        )
    manifests = manifests_after_search

if len(manifests) > 1:
    paths = ", ".join(str(m) for m in manifests)
    raise RuntimeError(
        f"\n"
        f"Multiple cloud batches found: {paths}\n"
        f"Remove all but the one you want to process and run again."
    )

BATCH_DIR = manifests[0].parent if manifests[0].suffix == ".json" else manifests[0]

# Copy from read-only Kaggle input to writable working directory
WORK_DIR = Path(WORK_DIR_PATH)
WORK_DIR.mkdir(parents=True, exist_ok=True)
shutil.copytree(str(BATCH_DIR), str(WORK_DIR), dirs_exist_ok=True)

required = ["manifest.json", "input.jsonl", "worker.py", "requirements.txt"]
missing = [f for f in required if not (WORK_DIR / f).exists()]
if missing:
    raise FileNotFoundError(
        f"Bundle is incomplete — missing: {missing}\n"
        f"Re-export the batch locally and re-add the dataset."
    )

import json
manifest = json.loads((WORK_DIR / "manifest.json").read_text(encoding="utf-8"))
BATCH_ID = manifest["batch_id"]
MODEL = manifest["model"]
RULE_COUNT = manifest["rule_count"]
RESULT_FILE = WORK_DIR / f"{BATCH_ID}-results.jsonl"
ERRORS_FILE = WORK_DIR / "errors.jsonl"

print(f"Batch:       {BATCH_ID}")
print(f"Model:       {MODEL}")
print(f"Rules:       {RULE_COUNT}")
print(f"Work dir:    {WORK_DIR}")
print(f"Result file: {RESULT_FILE}")
```

```python
# ============================================================
# CELL 3 — INSTALL DEPENDENCIES
# ============================================================
import shutil as _shutil

print("Installing Python dependencies...")
subprocess.run(
    [sys.executable, "-m", "pip", "install", "-q", "-r", str(WORK_DIR / "requirements.txt")],
    check=True
)

if not _shutil.which("zstd"):
    print("Installing zstd...")
    subprocess.run(["apt-get", "install", "-y", "-q", "zstd"], check=False)

print("Dependencies ready.")
```

```python
# ============================================================
# CELL 4 — INSTALL OLLAMA (if missing)
# Reuses existing installation when available.
# ============================================================
import shutil as _shutil

if _shutil.which("ollama"):
    print("Ollama already installed:", _shutil.which("ollama"))
else:
    print("Installing Ollama...")
    result = subprocess.run(
        ["bash", "-lc", "curl -fsSL https://ollama.com/install.sh | sh"],
        capture_output=True, text=True
    )
    if result.returncode != 0:
        print(result.stderr[-2000:])
        raise RuntimeError("Ollama installation failed. Check that Internet access is enabled in Kaggle settings.")
    print("Ollama installed.")
```

```python
# ============================================================
# CELL 5 — START OLLAMA SERVER
# Reuses a running server; starts one if not yet running.
# Bounded readiness loop — does NOT blindly sleep.
# ============================================================
import requests, time

def ollama_ready():
    try:
        return requests.get(f"{OLLAMA_URL}/api/tags", timeout=3).ok
    except requests.RequestException:
        return False

if ollama_ready():
    print("Ollama server already running.")
else:
    print("Starting Ollama server...")
    ollama_log = (WORK_DIR / "ollama.log").open("ab")
    _server_proc = subprocess.Popen(
        ["ollama", "serve"],
        stdout=ollama_log, stderr=subprocess.STDOUT
    )
    for attempt in range(60):
        time.sleep(1)
        if ollama_ready():
            print(f"Ollama ready after {attempt + 1}s.")
            break
    else:
        log_tail = (WORK_DIR / "ollama.log").read_text(errors="replace")[-2000:]
        print(log_tail)
        raise RuntimeError(
            "Ollama did not become ready at 127.0.0.1:11434 within 60 seconds.\n"
            "See ollama.log above for details."
        )
```

```python
# ============================================================
# CELL 6 — PULL QWEN MODEL (if not already present)
# Checks local model list first. Pulls only if missing.
# ============================================================
import requests

def model_exists(model_name):
    try:
        response = requests.get(f"{OLLAMA_URL}/api/tags", timeout=10)
        response.raise_for_status()
        models = response.json().get("models", [])
        return any(
            entry.get("name") == model_name or entry.get("model") == model_name
            for entry in models
        )
    except requests.RequestException:
        return False

if model_exists(MODEL):
    print(f"Model already present: {MODEL}")
else:
    print(f"Pulling model: {MODEL} (this may take several minutes on first run)...")
    result = subprocess.run(["ollama", "pull", MODEL], capture_output=True, text=True)
    if result.returncode != 0:
        print(result.stderr[-2000:])
        raise RuntimeError(
            f"Model download failed: {MODEL}\n"
            "Check that Internet access is enabled. Input files and reservations are unchanged."
        )
    print(f"Model ready: {MODEL}")

import time
time.sleep(2)
smi = subprocess.run(
    ["nvidia-smi", "--query-gpu=name,memory.used,memory.free", "--format=csv,noheader"],
    capture_output=True, text=True
)
if smi.returncode == 0:
    print(f"GPU state: {smi.stdout.strip()}")
```

```python
# ============================================================
# CELL 7 — RUN WORKER
# Fully automatic. No user edits needed.
# Worker skips already-completed (rule_id, context_sha256) pairs.
# ============================================================

command = [
    sys.executable, "-u", str(WORK_DIR / "worker.py"),
    "--manifest",           str(WORK_DIR / "manifest.json"),
    "--input",              str(WORK_DIR / "input.jsonl"),
    "--output",             str(RESULT_FILE),
    "--errors",             str(ERRORS_FILE),
    "--environment",        "kaggle",
    "--ollama-url",         OLLAMA_URL,
    "--retries",            str(RETRIES),
    "--stop-after-errors",  str(STOP_AFTER_ERRORS),
]

print("Starting worker...")
print(" ".join(command))
print("-" * 60)

completed_proc = subprocess.run(command)
print(f"\nWorker exit code: {completed_proc.returncode}")
```

```python
# ============================================================
# CELL 8 — COMPLETION SUMMARY
# ============================================================
result_count = 0
error_count = 0

if RESULT_FILE.exists():
    result_count = sum(1 for line in RESULT_FILE.read_bytes().splitlines() if line.strip())
if ERRORS_FILE.exists():
    error_count = sum(1 for line in ERRORS_FILE.read_bytes().splitlines() if line.strip())

print()
print("=" * 49)
print("KAGGLE BATCH COMPLETE")
print("=" * 49)
print(f"Batch:   {BATCH_ID}")
print(f"Results: {result_count}")
print(f"Errors:  {error_count}")
print(f"Output:  {RESULT_FILE}")
print()
print("NEXT STEPS:")
print(f"  1. Download:  {RESULT_FILE.name}")
print(f"     (Kaggle output panel -> click the file)")
print(f"  2. Place it in:  data\\cloud_results\\  (on your local machine)")
print(f"  3. Run locally:  .\\cloud-import.ps1")

try:
    from IPython.display import FileLink, display
    display(FileLink(str(RESULT_FILE)))
    if error_count > 0:
        display(FileLink(str(ERRORS_FILE)))
except Exception:
    pass
```
